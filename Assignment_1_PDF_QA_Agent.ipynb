```python
from pdf2image import convert_from_path

pages = convert_from_path(
    "Web and Mobile App Report.pdf",
    dpi=150
)

print("Pages converted:", len(pages))
```

```python
import pytesseract

all_text = ""

for i, page in enumerate(pages):
    page_text = pytesseract.image_to_string(page)
    all_text += f"\n--- Page {i+1} ---\n"
    all_text += page_text

print("OCR completed.")
print("Characters extracted:", len(all_text))
print(all_text[:3000])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

chunks = splitter.split_text(all_text)

print("Number of chunks:", len(chunks))
print("\nFirst chunk:\n")
print(chunks[0])
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_community.vectorstores import FAISS

embeddings = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001",
    google_api_key=GEMINI_API_KEY
)

vector_db = FAISS.from_texts(
    chunks,
    embeddings
)

print("Vector database created successfully.")
```

```python
retriever = vector_db.as_retriever(
    search_kwargs={"k": 3}
)

print("Retriever created successfully.")
```

```python
def ask_pdf(question):

    # Retrieve relevant chunks
    docs = retriever.invoke(question)

    # Combine retrieved content
    context = "\n\n".join(
        doc.page_content for doc in docs
    )

    # Create prompt
    prompt = f"""
You are a PDF question-answering agent.

Answer the user's question using ONLY the information
provided in the context.

If the answer is not available in the context, say:
"Information not found in the document."

Context:
{context}

Question:
{question}

Give a clear and concise answer.
"""

    # Ask Gemini
    response = llm.invoke(prompt)

    return response.content
```

```python
question = "What is the purpose of the Leave Management System?"

answer = ask_pdf(question)

print("Question:", question)
print("\nAnswer:", answer)
```

```python
def ask_pdf(question):

    docs = retriever.invoke(question)

    context = "\n\n".join(
        doc.page_content for doc in docs
    )

    prompt = f"""
Answer the question using only the provided context.

Context:
{context}

Question:
{question}

If the answer is not found, say:
"Information not found in the document."

Give a clear and concise answer.
"""

    response = llm.invoke(prompt)

    return response.content
```
